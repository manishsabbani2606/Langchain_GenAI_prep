```python
from langchain_text_splitters import HTMLHeaderTextSplitter
html_string = """
<html>
  <body>
    <h1>Main Title</h1>
    <p>This is the introduction paragraph.</p>      
    <h2>Subsection</h2>
    <p>This is a paragraph under the subsection.</p>
    </body>
</html>
"""
header_splitter= [(
    "h1","Header 1"),
    ("h2","Header 2"),
    ("h3","Header 3")
]

text_splitter = HTMLHeaderTextSplitter(headers_to_split_on=header_splitter)
final_docs = text_splitter.split_text(html_string)
final_docs
```

### Splitting the JSON DATA using json splitter

```python
import json
import requests
```

```python
json_data = requests.get("https://api.github.com/repos/langchain-ai/langchain-community").json()
```

```python
json_data
```

```python
from langchain_text_splitters import RecursiveJsonSplitter
```

```python
json_splitter = RecursiveJsonSplitter(max_chunk_size=300)
json_chunks = json_splitter.split_json(json_data)
```

```python
for chunk in json_chunks[:5]:
    print(chunk)
   
```

```python
#The splitter can also output in different formats like CSV, XML, etc.
docs = json_splitter.create_documents([json_data])
for doc in docs[:5]:
    print(doc)
```

```python
csv_doc = json_splitter.split_text(json_data)
```

```python
print(csv_doc[0])
```
