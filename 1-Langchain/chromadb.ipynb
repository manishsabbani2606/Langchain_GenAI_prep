#### Chroma DB vector Store

```python
### We have to install chromadb package in our requirements.txt file to use chromadb vector store
#### we also have the langchain_chromadb package to use chromadb with langchain
```

```python
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter,RecursiveCharacterTextSplitter
```

```python
loader = TextLoader('speech2.txt')
```

```python
data = loader.load()
data
```

```python
#spliting the data 
splits = RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=0).split_documents(data)
splits
```

```python
embeddings = OllamaEmbeddings(model = "gemma:2b")
vectorstore = Chroma.from_documents(splits, embeddings)
vectorstore
```

```python
query = "how many titles won by CSK?"
docs = vectorstore.similarity_search(query)
docs
docs[0].page_content
```

```python
# SAVING TO THE DISK
vectordb = Chroma.from_documents(splits, embeddings, persist_directory="./chromadb_index")
```

```python
score = vectordb.similarity_search_with_score(query)
score
```

```python
retriver = vectordb.as_retriever()
retriver.invoke(query)
```
