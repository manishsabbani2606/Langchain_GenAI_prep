### Prompt Template

```python
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that answer everything."),
    ("user", "Help me out with the SQL Query for each situation {text}"),
])
```

```python
prompt
```

```python

#combine the prompt and llm to create a chain
chain = prompt|llm
chain.invoke({"Give me the list of employees in the HR department."})
```

```python
response3= chain.invoke({"text":"Hello Manish how are you?"})
```

```python
response3
```

```python
type(response3)
```

```python
## String output parser
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
```

```python
chain= prompt|llm|output_parser
result = chain.invoke({"Give me the list of employees in the HR department."})
result
```
