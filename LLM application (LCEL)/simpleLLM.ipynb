### Building a simple LLM Application with LCEL

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
import openai
groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
pip install langchain-groq
```

```python
from langchain_groq import ChatGroq
from langchain_openai import ChatOpenAI
```

```python
model=ChatGroq(model="openai/gpt-oss-120b",groq_api_key=groq_api_key)
```

#### LLM Messages (langchain messages)

```python
#Language translation example
from langchain_core.messages import HumanMessage, SystemMessage
messages= [
    SystemMessage(content="Translate the following from english to hindi"),
    HumanMessage(content="Hello, how are you?")
]

result=model.invoke(messages)
```

```python
# String out parsing with langchain
from langchain_core.output_parsers import StrOutputParser
parser = StrOutputParser()
parser.invoke(result)
```

```python
### Using LCEL with Groq LLM

chain = model|parser
chain.invoke(messages)
```

```python
### Prompt Templates
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate

generictemplate = "Transalate the following from {source_language} to {target_language}: {text}"
prompt = ChatPromptTemplate.from_messages([
    ("system",generictemplate),
    ("user","{text}")
])


prompt.invoke({"source_language":"english","target_language":"hindi","text":"Hello, how are you?"})
```

```python
output=chain.invoke(prompt.invoke({"source_language":"english","target_language":"TELUGU","text":"""The morning sky stretched quietly above the city as traffic lights blinked from red to green. Somewhere between routine and ambition, people carried their plans forward, one step at a time

Technology moves fast, but clarity moves faster. A simple idea, clearly explained, often outperforms a complex solution that no one understands."""}))
```
